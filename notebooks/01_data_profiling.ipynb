```python
import os
import pandas as pd

# Define raw data directory path
RAW_DATA_PATH = "../data/raw/"

# Retrieve all CSV files from the raw data directory
csv_files = [f for f in os.listdir(RAW_DATA_PATH) if f.endswith('.csv')]
print(f"Total CSV Files Found: {len(csv_files)}\n")

# Iterate through each file and inspect basic properties
for file in sorted(csv_files):
    df = pd.read_csv(os.path.join(RAW_DATA_PATH, file))
    print(f"File: {file}")
    print(f"   - Rows: {df.shape[0]:,}, Columns: {df.shape[1]}")
    print(f"   - Sample Columns: {list(df.columns[:3])}...\n")
```

```python
# Generate a data summary matrix for all datasets
summary_data = []

for file in sorted(csv_files):
    df = pd.read_csv(os.path.join(RAW_DATA_PATH, file))
    summary_data.append({
        "File Name": file,
        "Total Rows": df.shape[0],
        "Total Columns": df.shape[1],
        "Duplicate Rows": df.duplicated().sum(),
        "Columns with Nulls": sum(df.isnull().sum() > 0)
    })

# Convert summary list into a structured DataFrame
summary_df = pd.DataFrame(summary_data)
summary_df
```

```python
# Detailed null value analysis across all datasets
null_summary = []

for file in sorted(csv_files):
    df = pd.read_csv(os.path.join(RAW_DATA_PATH, file))
    null_counts = df.isnull().sum()
    null_cols = null_counts[null_counts > 0]
    
    for col, count in null_cols.items():
        null_summary.append({
            "File Name": file,
            "Column Name": col,
            "Missing Count": count,
            "Missing Percentage (%)": round((count / len(df)) * 100, 2)
        })

# Display missing values breakdown
null_df = pd.DataFrame(null_summary)
null_df
```
